# Multi-variable linear regression

hypothesis = H(x) = Wx + b

=> H(x1, x2, x3) => w1x1 + w2x2 + w3x3 + b

### Matrix multiplication(메트릭스 곱셈)

Hypothesis using matrix

(x1, x2, x3) * (w1  = (w1x1 + w2x2 + w3x3)
                w2
                w3)

### WX vs XW

- Lecture(theory) : H(x) = Wx + b
- Implementation(TensorFlow) : H(X) = XW

```python
import tensorflow as tf

x1_data = [73., 93., 89., 96., 73.]
x2_data = [80., 88., 91., 98., 66.]
x3_data = [75., 93., 90., 100., 70.]
y_data = [152., 185., 180., 196., 142.]

w1 = tf.Variable(tf.random.normal([1]), name='w1')
w2 = tf.Variable(tf.random.normal([1]), name='w2')
w3 = tf.Variable(tf.random.normal([1]), name='w3')
b = tf.Variable(tf.random.normal([1]), name='bias')

optimizer = tf.keras.optimizers.SGD(learning_rate=1e-5)

for step in range(2001):
    with tf.GradientTape() as tape:
        hypothesis = x1_data * w1 + x2_data * w2 + x3_data * w3 + b
        cost = tf.reduce_mean(tf.square(hypothesis - y_data))

    grads = tape.gradient(cost, [w1, w2, w3, b])
    optimizer.apply_gradients(zip(grads, [w1, w2, w3, b]))

    if step % 500 == 0:
        print(f"Step: {step:4d} | Cost: {cost.numpy():.4f}")


        print(f"  [학습된 파라미터]")
        print(f"  w1: {w1.numpy()[0]:.3f}, w2: {w2.numpy()[0]:.3f}, w3: {w3.numpy()[0]:.3f}, b: {b.numpy()[0]:.3f}")


        print(f"  [성능 비교]")
        print(f"  실제값(Y): {y_data}")

        preds = [round(x, 1) for x in hypothesis.numpy()]
        print(f"  예측값(H): {preds}")
        print("-" * 60)

```

```python
import tensorflow as tf
import numpy as np
import ast

xy = np.loadtxt('data-01-test-score.csv', delimiter=',', dtype=np.float32)
x_data = xy[:,0:-1]
y_data = xy[:,[-1]]

w = tf.Variable(tf.random.normal([3,1]), name='weight')
b = tf.Variable(tf.random.normal([1]), name='bias')

optimizer = tf.keras.optimizers.SGD(learning_rate=1e-5)

for step in range(2001):
    with tf.GradientTape() as tape:
        hypothesis = tf.matmul(x_data, w) + b
        cost = tf.reduce_mean(tf.square(hypothesis - y_data))

    grads = tape.gradient(cost, [w, b])
    optimizer.apply_gradients(zip(grads, [w, b]))

    if step % 100 == 0:
        print(f"Step: {step:4d} | Cost: {cost.numpy():.4f}")
print("--- 학습 완료! ---", )

while True:
    print("\n점수 데이터를 리스트 형태로 입력하세요. (종료하려면 q 입력)")
    print("예시 : [[73,80,75]] 또는 [[73,80,87],[93,88,93]]")

    user_input= input("입력 > ")

    if user_input == 'q':
        break

    try:
        my_scores = ast.literal_eval(user_input)

        my_tensor = tf.constant(my_scores, dtype=tf.float32)

        pred_result = tf.matmul(my_tensor, w) + b

        print("-" * 30)
        for i, res in enumerate(pred_result.numpy()):
            print(f"데이터 {i+1}번 {my_scores[i]}의 예상 점수: {res[0]:.2f}점")
        print("-" * 30)
    except ValueError:
        print("⚠ 형식이 잘못되었습니다. 괄호[]와 콤마, 숫자를 확인해주세요.")
    except Exception as e:
        print(f"⚠ 오류 발생: {e}")
```
