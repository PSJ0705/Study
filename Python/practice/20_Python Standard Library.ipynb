##  표준 라이브러리

파이썬 표준 라이브러리는 파이썬을 설치할 때 자동으로 컴퓨터에 설치된다.

sys, re 모듈은 파이썬의 중요한 표준 라이브러리이다.

1. datetime. date는 연,월, 일로 날짜를 표현할 때 사용하는 함수이다.

```python
import datetime

day1 = datetime.date(2021, 12, 14)
day2 = datetime.date(2023, 4, 5)

print(day1)
print(day2)

# 두 날짜의 차이는 다음과 같이 뺄셈으로 쉽게 구할 수 있다.

diff = day2 - day1
diff.days
```

요일은 datetime.date 객체의 weekday 함수를 사용하면 쉽게 구할 수 있다.

```python
day = datetime.date(2026, 2, 7)
day.weekday()

# 0은 월요일, 1은 화요일, 2는 수요일,..., 6은 일요일이 된다.
```

```python
# 월요일은 1, 화요일은 2, ..., 일요일은 7을 반환하게 하려면 isoweekday 함수를 하용하면 된다.

day.isoweekday()
```

2. Time : 시간과 관련된 time 모듈에는 함수가 매우 많다. 그 중 가장 유용한 몇 가지만 알아보자

2-2. time.time() 은 UTC를 사용하여 현재 시간을 실수 형태로 반환하는 함수이다. 1970년 1월 1일 0시 0분 0초를 기준으로 지난 시간을 초 단위로 반환한다.

```python
import time

time.time()
```

2-3. time.asctime은 time.localtime이 반환한 튜플 형태의 값을 인수로 받아서 날짜와 시간을 알아보기 쉬운 형태로 반환하는 함수이다.

```python
time.asctime(time.localtime(time.time()))
```

2-4. time.ctime(time.localtime(time.time)))은 간단한게 time.ctime()으로 표시할 수 있다. ctime이 asctime과 다른 점은 항상 현재 시간만을 반환한다는 점이다.

```python
time.ctime()
```

2-5. time.strftime 함수는 시간에 관계된 것을 세밀하게 표현하는 여러 가지 포맷 코드를 제공한다.

```python
time.strftime('%A', time.localtime(time.time()))
```

2-6. time.sleep 함수는 주로 루프 안에서 많이 사용한다. 이 함수를 사용하면 일정한 시간 간격을 두고 루프를 실행할 수 있다.

```python
import time

for i in range(10):
    print(i, time.ctime())
    time.sleep(1)
```

4-1. random은 난수를 발생시키는 모듈이다.

다음은 0.0에서 1.0 사이의 실수 중에서 난수 값을 반환하는 예이다.

```python
import random
random.random()
```

다음 예에서는 1에서 10 사이의 정수 중에서 난수 값을 반환한다.

```python
random.randint(1, 10)
```

 다음 예에서는 1에서 55 사이의 정수 중에서 난수 값을 반환한다.

```python
random.randint(1, 55)
```

```python
# random 모듈을 이용한 예시 함수

import random

def random_pop(data):
    number = random.randint(0, len(data)-1)
    return data.pop(number)

if __name__ == "__main__":
    data = [1,2,3,4,5]
    while data:
        print(random_pop(data))
```

```python
# random.choice 함수를 이용한 버전

import random

def random_pop(data):
    number = random.choice(data)
    data.remove(number)
    return number

if __name__ == "__main__":
    data = [1,2,3,4,5]
    while data:
        print(random_pop(data))
```

```python
# 리스트의 항목들을 무작위로 섞고 싶을 때에는 random.sample 함수를 사용하면 된다.

import random
data = [1,2,3,4,5]
random.sample(data, len(data))      # len(data)는 무작위로 섞을 데이터의 개수를 의미한다.
```
