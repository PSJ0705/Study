2. all(X)는 반복 가능한 데이터 X를 입력값으로 받으며 이 x의 요소가 모두 참이면 True, 거짓이 하나라도 있으면 False를 반환한다.

```python
print(all([1,2,3]))

print(all([0]))     # 0값은 거짓이므로 Flase를 반환

print(all([]))      # all의 입력 인수가 빈 값인 경우에는 True를 반환
```

7. enumerate는 '열거하다'라는 뜻이다. 이 함수는 순서가 있는 데이터(리스트, 튜플, 문자열)를 입력으로 받아 인덱스 값을 포함하는 enumerate 객체를 반환한다.

```python
for i, name in enumerate(['body', 'foo', 'bar']):
    print(i, name)
```

인덱스 값과 함께 body, foo, bar가 순서대로 출력되었다. 즉, enumerate를 for문과 함게 사용하면 자료형의 현재 순서 index와 그 값을 쉽게 알 수 있다.

for 문처럼 반복되는 구간에서 객체가 현재 어느 위치에 있는지 알려주는 인덱스 값이 필요할 때, enumerate함수를 사용하면 매우 유용하다.

9. filter란 '무엇인가를 걸러낸다'라는 뜻으로, filter함수도 이와 비슷한 기능을 한다

        filter(함수, 반복 가능한 데이터)


filter 함수는 첫 번째 인수로 함수, 두 번째 인수로 그 함수에 차례로 들어갈 반복 가능한 데이터를 받는다. 그리고 반복 가능한 데이터의 요소 순서대로 함수를 호출했을 때 반환값이 참인 것만 묶어서(걸러내서)반환한다.

```python
def positive(l):
    result = []
    for i in l:
        if i > 0:
            result.append(i)
    return result

print(positive([1,2,3,-2,0,-5,6]))
```

위와 같은 함수는 리스트의 입력으로 받아 각각의 요소를 판별해 양수값만 반환하는 함수이다.

filter함수를 사용하면 위 내용을 다음과 같이 간단하게 작성할 수 있다.

```python
def positive(x):
    return x > 0

print(list(filter(positive, [1,2,3,-2,0,-5,6])))
```

이 예제는 lambda를 사용하면 더욱 간단해진다.

```python
print(list(filter(lambda x: x > 0, [1,2,3,-2,0,-5,6])))
```

14. isinstance(object, class)함수는 첫 번재 인수로 객체, 두 번재 인수로 클래스를 받는다. 입력으로 받은 객체가 그 클래스의 인스턴스인지를 판단하여 참이면 True, 거짓이면 False를 반환한다.

```python
class Person: pass

a = Person()

print(isinstance(a, Person))

b = 3

print(isinstance(b, Person))
```

17. map(f, iterable)은 함수(f)와 반복 가능한 데이터를 입력으로 받는다. map은 입력받은 데이터의 각 요소에 함수 f를 적용한 결과를 반환하는 함수이다.

```python
def two_times(x):
    return x * 2

print(list(map(two_times, [1,2,3])))
```

```python
print(list(map(lambda x: x*2, [1,2,3])))
```
